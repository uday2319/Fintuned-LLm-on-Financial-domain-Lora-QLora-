```python
from peft import LoraConfig,get_peft_model,prepare_model_for_kbit_training
from datasets import Dataset
import torch
from transformers import AutoModelForCausalLM ,AutoTokenizer,BitsAndBytesConfig
model_name="TinyLlama/TinyLlama-1.1B-Chat-v1.0"

bnb_config=BitsAndBytesConfig(
     load_in_4bit=True,
     bnb_4bit_compute_dtype=torch.float16,
     bnb_4bit_use_double_quant=True,
     bnb_4bit_quant_type="nf4"
 )
tokenizer=AutoTokenizer.from_pretrained(model_name)
model=AutoModelForCausalLM.from_pretrained(model_name,
                                           quantization_config=bnb_config,device_map="auto")

tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.eos_token_id

model=prepare_model_for_kbit_training(model)
example = [
{"instruction": "What is inflation?",
 "response": "Definition: Inflation is the increase in prices over time.\nExample: Milk price rising from ₹50 to ₹60.\nImpact: Reduces purchasing power.\nWhen to worry: When income does not increase with prices."},

{"instruction": "What is SIP?",
 "response": "Definition: SIP is investing a fixed amount regularly.\nExample: Investing ₹1000 monthly in mutual funds.\nImpact: Helps build wealth over time.\nWhen to worry: During long market downturns."},

{"instruction": "What is a mutual fund?",
 "response": "Definition: A mutual fund pools money to invest in assets.\nExample: Investing ₹5000 in an equity fund.\nImpact: Provides diversification.\nWhen to worry: Market fluctuations affect returns."},

{"instruction": "What is a stock?",
 "response": "Definition: A stock represents ownership in a company.\nExample: Buying shares of Infosys.\nImpact: Potential for high returns.\nWhen to worry: High market volatility."},

{"instruction": "What is a bond?",
 "response": "Definition: A bond is a loan given to a government or company.\nExample: Government bond paying fixed interest.\nImpact: Provides stable income.\nWhen to worry: Rising interest rates reduce value."},

{"instruction": "What is diversification?",
 "response": "Definition: Diversification means spreading investments.\nExample: Investing in stocks, gold, and bonds.\nImpact: Reduces overall risk.\nWhen to worry: Over-diversification lowers returns."},

{"instruction": "What is budgeting?",
 "response": "Definition: Budgeting is planning income and expenses.\nExample: Allocating ₹10,000 for rent.\nImpact: Helps control spending.\nWhen to worry: When expenses exceed income."},

{"instruction": "What is an emergency fund?",
 "response": "Definition: Emergency fund is savings for unexpected situations.\nExample: Saving 3–6 months of expenses.\nImpact: Provides financial security.\nWhen to worry: If savings are insufficient."},

{"instruction": "What is income tax?",
 "response": "Definition: Income tax is tax paid on earnings.\nExample: Salary taxed based on slabs.\nImpact: Reduces take-home income.\nWhen to worry: Poor tax planning increases liability."},

{"instruction": "What is GST?",
 "response": "Definition: GST is tax on goods and services.\nExample: Paying tax on purchases.\nImpact: Affects cost of products.\nWhen to worry: Increasing tax rates."},

{"instruction": "What is FD?",
 "response": "Definition: Fixed Deposit is a one-time investment with fixed returns.\nExample: ₹50,000 invested for 1 year.\nImpact: Safe but low returns.\nWhen to worry: Returns lower than inflation."},

{"instruction": "What is RD?",
 "response": "Definition: Recurring Deposit is monthly saving scheme.\nExample: ₹2000 deposited monthly.\nImpact: Builds disciplined saving habit.\nWhen to worry: Low interest rates."},

{"instruction": "What is compound interest?",
 "response": "Definition: Interest earned on principal and previous interest.\nExample: ₹1000 growing yearly.\nImpact: Accelerates wealth growth.\nWhen to worry: Works against you in loans."},

{"instruction": "What is liquidity?",
 "response": "Definition: Liquidity is how easily an asset converts to cash.\nExample: Cash is liquid, property is not.\nImpact: Helps in emergencies.\nWhen to worry: Too many illiquid assets."},

{"instruction": "What is net worth?",
 "response": "Definition: Net worth is assets minus liabilities.\nExample: ₹5 lakh assets minus ₹2 lakh debt.\nImpact: Measures financial health.\nWhen to worry: Increasing debt."},

{"instruction": "What is credit score?",
 "response": "Definition: Credit score shows creditworthiness.\nExample: CIBIL score.\nImpact: Affects loan approval.\nWhen to worry: Low score affects borrowing."},

{"instruction": "What is EMI?",
 "response": "Definition: EMI is fixed monthly loan payment.\nExample: ₹10,000 monthly loan repayment.\nImpact: Affects monthly budget.\nWhen to worry: High EMI burden."},

{"instruction": "What is savings account?",
 "response": "Definition: Bank account to store money safely.\nExample: Depositing salary.\nImpact: Provides liquidity.\nWhen to worry: Low interest rates."},

{"instruction": "What is financial planning?",
 "response": "Definition: Planning finances for future goals.\nExample: Saving for retirement.\nImpact: Ensures stability.\nWhen to worry: Lack of planning leads to stress."},

{"instruction": "What is insurance?",
 "response": "Definition: Protection against financial risks.\nExample: Health insurance policy.\nImpact: Reduces unexpected expenses.\nWhen to worry: No insurance coverage."},

{"instruction": "What is PPF?",
 "response": "Definition: Government-backed long-term savings scheme.\nExample: 15-year investment.\nImpact: Tax-free returns.\nWhen to worry: Long lock-in period."},

{"instruction": "What is NPS?",
 "response": "Definition: Retirement savings scheme.\nExample: Monthly pension contributions.\nImpact: Builds retirement corpus.\nWhen to worry: Market dependency."},

{"instruction": "What is gold investment?",
 "response": "Definition: Investing in gold assets.\nExample: Gold ETF.\nImpact: Hedge against inflation.\nWhen to worry: Price volatility."},

{"instruction": "What is real estate investment?",
 "response": "Definition: Investing in property.\nExample: Buying a house.\nImpact: Long-term wealth creation.\nWhen to worry: Low liquidity."},

{"instruction": "What is risk?",
 "response": "Definition: Chance of losing money.\nExample: Stock market fall.\nImpact: Affects returns.\nWhen to worry: High-risk exposure."},

{"instruction": "What is return?",
 "response": "Definition: Profit from investment.\nExample: Stock price increase.\nImpact: Builds wealth.\nWhen to worry: Low returns."},

{"instruction": "What is debt?",
 "response": "Definition: Borrowed money that must be repaid.\nExample: Personal loan.\nImpact: Financial obligation.\nWhen to worry: High interest rates."},

{"instruction": "What is passive income?",
 "response": "Definition: Income earned with little effort.\nExample: Rental income.\nImpact: Financial independence.\nWhen to worry: Low income streams."},

{"instruction": "What is expense tracking?",
 "response": "Definition: Monitoring spending habits.\nExample: Monthly expense log.\nImpact: Improves savings.\nWhen to worry: Overspending."}
]

dataset=Dataset.from_list(example)
lora_config=LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["q_proj","v_proj"],
    lora_dropout=0.1
)
model =get_peft_model(model,lora_config)

def format_data(example):
  text = f"### Instruction:\n{example['instruction']}\n\n### Response:\n{example['response']}"
  return {"text":text}

dataset=dataset.map(format_data)

def tokenize(example):
  tokens=tokenizer(example["text"],truncation=True,padding="max_length",max_length=512)
  tokens["labels"] = tokens["input_ids"].copy()
  return tokens

dataset=dataset.map(tokenize,batched=True)
from transformers import Trainer,TrainingArguments

training_args=TrainingArguments(
    output_dir="/.finetuned_model",
    per_device_train_batch_size=2,
    num_train_epochs=4,
    logging_steps=20,
    fp16=True,
    learning_rate = 1e-4,
    save_strategy="epoch"
)
trainer=Trainer(
    model=model,
    args=training_args,
    train_dataset=dataset
)

trainer.train()
```

```python
import torch

def generate_response(model, prompt):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            temperature=0.7,
            do_sample=False,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id

        )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
prompt = """### Instruction:
what is SIP? in the following format:
Definition:
Example:
Impact:
When to worry:

### Response:
"""
print(generate_response(model, prompt))
```
